**TUGAS**

Dataset: Wisconsin Breast Cancer (569 data, diagnosis M = Malignant / B = Benign)

1. Pisahkan antara variabel yang dapat digunakan dan yang tidak dapat digunakan
2. Lakukan proses encoding pada kolom `diagnosis`
3. Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik

# Langkah 0 - Load Library

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

# Langkah 1 - Load Data

```python
dpath = '/content/wbc.csv'
df = pd.read_csv(dpath)
df.head()
```

# Langkah 2 - Inspeksi Data

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

Hasil inspeksi:

- `id` bertipe int64 tetapi hanya berupa nomor identitas pasien, tidak memiliki hubungan dengan diagnosis
- `Unnamed: 32` memiliki 569 nilai kosong, artinya kolom tersebut kosong seluruhnya
- 31 kolom lainnya lengkap tanpa nilai kosong

Karena `Unnamed: 32` kosong 100%, kolom ini tidak diimputasi melainkan dibuang. Imputasi hanya masuk akal apabila masih ada sebagian data yang terisi sebagai dasar pengisian.

# Langkah 3 - Slice Data

```python
kolom_tidak_terpakai = ['id', 'Unnamed: 32']

df_tidak_terpakai = df[kolom_tidak_terpakai]   # variabel yang TIDAK dapat digunakan
df = df.drop(columns=kolom_tidak_terpakai)     # variabel yang dapat digunakan

print('Tidak dapat digunakan:', list(df_tidak_terpakai.columns))
print('Dapat digunakan     :', list(df.columns))
```

# Langkah 4 - Verifikasi Hasil Slice

```python
df.shape
```

```python
df.head()
```

# Langkah 5 - Encoding

```python
print(df['diagnosis'].value_counts())   # sebelum encoding

le = LabelEncoder()                       # membuat objek dari LabelEncoder
df['diagnosis'] = le.fit_transform(df['diagnosis'])   # proses encoding

print(dict(zip(le.classes_, le.transform(le.classes_))))   # mapping hasil encoding
```

LabelEncoder mengurutkan label secara alfabet, sehingga `B` (Benign) menjadi 0 dan `M` (Malignant) menjadi 1.

# Langkah 6 - Verifikasi Hasil Encoding

```python
df['diagnosis'].value_counts()
```

```python
df.head()
```

# Langkah 7 - Standarisasi

```python
kolom_numerik = df.drop(columns=['diagnosis']).columns   # 30 kolom fitur numerik

std = StandardScaler()                                    # membuat objek dari StandardScaler
df[kolom_numerik] = std.fit_transform(df[kolom_numerik])  # proses standarisasi

print('Jumlah kolom yang distandarisasi:', len(kolom_numerik))
```

Kolom `diagnosis` tidak ikut distandarisasi karena merupakan target (label), bukan fitur.

# Langkah 8 - Verifikasi Hasil Standarisasi

```python
df.head()
```

```python
df[kolom_numerik].describe()
```

```python
# Setelah standarisasi, setiap kolom numerik memiliki mean = 0 dan standar deviasi = 1
print('Mean :', df[kolom_numerik].mean().round(6).unique())
print('Std  :', df[kolom_numerik].std(ddof=0).round(6).unique())
```

# Langkah 9 - Simpan Hasil

```python
df.to_csv('wbc-fixed.csv', index=False)
df.shape
```
